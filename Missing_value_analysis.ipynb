```python
import pandas as pd

# Load Titanic dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# --------------------------------------------
# 1. Display basic information
# --------------------------------------------
print("=" * 60)
print("MISSING VALUE ANALYSIS - TITANIC DATASET")
print("=" * 60)

print("\nDataset Shape:")
print(df.shape)

print("\nFirst 5 Records:")
print(df.head())

# --------------------------------------------
# 2. Dataset information
# --------------------------------------------
print("\n" + "=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

df.info()

# --------------------------------------------
# 3. Count missing values in each column
# --------------------------------------------
missing_count = df.isnull().sum()

print("\n" + "=" * 60)
print("MISSING VALUES IN EACH COLUMN")
print("=" * 60)

print(missing_count)

# --------------------------------------------
# 4. Calculate missing-value percentage
# --------------------------------------------
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_count,
    "Missing Percentage": missing_percentage.round(2)
})

print("\n" + "=" * 60)
print("MISSING VALUE SUMMARY")
print("=" * 60)

print(missing_summary)

# --------------------------------------------
# 5. Show only columns having missing values
# --------------------------------------------
columns_with_missing = missing_summary[
    missing_summary["Missing Values"] > 0
]

print("\n" + "=" * 60)
print("COLUMNS REQUIRING MISSING-VALUE TREATMENT")
print("=" * 60)

print(columns_with_missing)

# --------------------------------------------
# 6. Total number of missing values
# --------------------------------------------
total_missing = df.isnull().sum().sum()

print("\n" + "=" * 60)
print("TOTAL MISSING VALUES")
print("=" * 60)

print("Total missing values in dataset:", total_missing)

# --------------------------------------------
# 7. Identify columns based on missing percentage
# --------------------------------------------
print("\n" + "=" * 60)
print("INTERPRETATION")
print("=" * 60)

for column in columns_with_missing.index:
    percentage = missing_percentage[column]

    if percentage > 50:
        print(
            f"{column}: {percentage:.2f}% missing - "
            "High missing values; consider dropping the column "
            "or applying a suitable treatment."
        )
    elif percentage > 20:
        print(
            f"{column}: {percentage:.2f}% missing - "
            "Moderate missing values; treatment is required."
        )
    else:
        print(
            f"{column}: {percentage:.2f}% missing - "
            "Low missing values; imputation may be appropriate."
        )

```
